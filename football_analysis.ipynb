***Data Cleaning***

```python
df.isnull()
```

```python
df.drop(['Home_Team_Yellowcard','Away_Team_Yellowcard','Home_Team_Redcard','Away_Team_Redcard','Home_Team_Passes','Away_Team_Passes','Home_Team_Offside','Away_Team_Offside','Home_Team_Corner','Away_Team_Corner'],axis = 1,inplace = True)
```

```python
df['Total_Goals'] = df['Home_Team_Score'] + df['Away_Team_Score']
```

```python
df['Goal_diff'] = abs(df['Home_Team_Score'] - df['Away_Team_Score'])
```

```python
df['Winning_Team'] = np.where(df['Home_Team_Score'] > df['Away_Team_Score'], df['Home_Team'],
np.where(df['Home_Team_Score'] < df['Away_Team_Score'], df['Away_Team'],'Draw'))
```

```python
plt.figure(figsize=(20, 5)) 
plt.subplot(1,2,1)
x = df['ID']
y = df['Away_Team_Possession']
plt.bar(x, y, color = 'red', label = 'Away_Team')
plt.xlabel('ID')
plt.ylabel('Possession (%)')
plt.title('Away Team Possession')
plt.subplot(1,2,2)
x1 = df['ID']
y1 = df['Home_Team_Possession']
plt.bar(x1, y1, color = 'blue')
plt.xlabel('ID')
plt.ylabel('Possession (%)')
plt.title('Away Team Possession')
```

***Insights & Conclusions***

```python
team_stats = df.groupby('Winning_Team').agg({
    'Total_Goals': 'sum',
    'Goal_diff': 'mean',
    'Home_Team_Shots': 'sum',
    'Home_Team_Possession': 'mean'
}).reset_index()
```

```python
team_stats.head(10)
```

```python
df_no_draws = df[df['Winning_Team'] != 'Draw']
df[df['Winning_Team'] != 'Draw']['Winning_Team'].value_counts().head(10).plot(kind='bar', color='green')
plt.title('Top 10 Winning Teams')
plt.xlabel('Team')
plt.ylabel('Wins')
plt.show()
```

```python
plt.bar(['Home Goals', 'Away Goals'], [df['Home_Team_Score'].sum(), df['Away_Team_Score'].sum()])
plt.title('Total Home vs Away Goals')

```
